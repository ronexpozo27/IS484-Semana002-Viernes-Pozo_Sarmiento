## 4.2. Representación de un problema mediante estados

### Parte A: Ficha del problema

| Estado | Descripción |
|---|---|
| **Estado inicial** | Almacén |
| **Estado objetivo** | Zona Norte |
| **Acciones posibles** | Desplazarse desde un punto de la ciudad hacia otro punto conectado directamente en el mapa. |
| **Costo de cada acción** | 1 paso por cada desplazamiento entre dos puntos conectados. |

### Parte B: Representación del mapa como grafo

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"]
}

print(mapa)
```

## 4.3. Implementación de BFS

Se implementa el algoritmo de búsqueda en amplitud (BFS) utilizando una cola. El algoritmo explora los nodos nivel por nivel hasta encontrar un camino desde el Almacén hasta Zona Norte.

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]  # Inicializo la cola con el nodo inicial
    visitados = set()  # Uso set para evitar visitar nodos repetidos

    while cola:
        camino = cola.pop(0)  # Extraigo el primer camino de la cola
        nodo_actual = camino[-1]  # Obtengo el ultimo nodo del camino

        if nodo_actual == objetivo:
            return camino  # Retorno el camino si llegue al objetivo

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)  # Marco el nodo como visitado

            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]  # Creo un nuevo camino
                cola.append(nuevo_camino)  # Agrego el camino al final de la cola

    return None  # No se encontro un camino


# Ejecuto BFS desde Almacen hasta ZonaNorte
resultado_bfs = bfs(mapa, "Almacen", "ZonaNorte")

print("Camino encontrado (BFS):", resultado_bfs)
print("Numero de pasos:", len(resultado_bfs) - 1)
```

## 4.4. Vectores y matrices con NumPy

### Parte A: Creación de vectores y matriz

Se representan cuatro observaciones mediante vectores de NumPy. Cada observación contiene dos características: temperatura y humedad. Los vectores se agrupan en una matriz para representar el conjunto de datos.

```python
import numpy as np

# Creo las observaciones con temperatura y humedad
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

# Agrupo las observaciones en una matriz
dataset = np.array([obs1, obs2, obs3, obs4])

# Muestro el dataset y sus dimensiones
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

### Parte B: Acceso a filas y columnas

```python
print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

## 4.5. Producto escalar y distancia euclidiana

### Parte A: Producto escalar

Se calcula el producto escalar entre las observaciones `obs1` y `obs2` utilizando la función `np.dot()` de NumPy.

```python
# Calculo el producto escalar aplicando la formula manual
producto_manual = np.sum(obs1 * obs2)
print("Producto escalar manual:", producto_manual)

# Calculo el mismo producto escalar usando NumPy
producto_numpy = np.dot(obs1, obs2)
print("Producto escalar con NumPy:", producto_numpy)
```

### Parte B: Distancia euclidiana

Se calcula la distancia euclidiana entre `obs1` y `obs2` mediante la fórmula matemática y mediante la función `np.linalg.norm()` de NumPy. Ambos métodos deben producir el mismo resultado.

```python
# Calculo la distancia aplicando la formula manual
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)

# Calculo la misma distancia usando NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```
